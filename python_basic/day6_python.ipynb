```python
# banking system
class Bank:
  bank_name = "NIC Asia"

  # constructor (called automatically when object is created)
  def __init__(self, name, address, balance, pin):
    self.name = name
    self.address = address
    self.balance = balance

  def deposit(self, amount):
    if not str(amount).isdigit() or amount <= 0:
      return "invalid amount"

    self.balance += amount
    return self.balance

  def withdraw(self, amount):
    if not str(amount).isdigit() or amount <= 0:
      return "invalid amount"

    if amount > self.balance:
      return "insufficient balance"

    self.balance -= amount
    return self.balance

  def check_balance(self):
    return self.balance
```

```python
class Person:
  def __init__(self,name,age,gender,address):
    self.name=name
    self.age=age
    self.gender=gender
    self.address=address

  def display_info(self):
    return f"The person name is {self.name}, age is {self.age}, gender is {self.gender} and lives in {self.address}"

class Employee(Person):
  company="ABC limited"
  def __init__(self,name,age,gender,address,salary,post,department):
    super().__init__(name,age,gender,address)
    self.salary=salary
    self.post=post
    self.department=department

  def display(self):
    super().display_info()
    return f"Salary is {self.salary}, post is {self.post} and department is {self.department}"

  def increase_salary(self,amount):
    self.salary += amount
    return self.salary

  def change_dept(self,new_dept):
    self.department=new_dept
    return new_dept

  def annual_salary(self):
    return self.salary*12



e1 = Employee("Hari", 25, "Male", "Pokhara", 50000, "Manager", "IT")

# print(e1.display_info())
print(e1.display())
print("Annual Salary:", e1.annual_salary())

e1.increase_salary(5000)
print("New Salary:", e1.salary)

e1.change_dept("HR")
print("New Department:", e1.department)

```

```python
import re
class Book:
    book_count = 0

    def __init__(self, id, title, author, price):
        self.id = id
        self.title = title
        self.author = author
        self.price = price
        self.is_available = True
        Book.book_count += 1

    def info(self):
        status = "Available" if self.is_available else "Borrowed"
        print(f"""
              Book ID   : {self.id}
              Title     : {self.title}
              Author    : {self.author}
              Price     : ${self.price}
              Status    : {status}
              """)

    def borrow(self):
        if self.is_available:
            self.is_available = False
            return True
        return False

    def return_book(self):
        self.is_available = True

    @classmethod
    def len_book(cls):
        return cls.book_count


class Member:
    member_count = 0

    def __init__(self, id, name, email):
        self.id = id
        self.name = name

        if self.is_valid_email(email):
            self.email = email
        else:
            raise ValueError("Invalid Email!")

        self.borrowed_books = []
        Member.member_count += 1

    def info(self):
        print(f"""
Member ID : {self.id}
Name      : {self.name}
Email     : {self.email}
Borrowed  : {len(self.borrowed_books)}
""")

    def borrow_book(self, book):
        self.borrowed_books.append(book)

    def return_book(self, book):
        if book in self.borrowed_books:
            self.borrowed_books.remove(book)

    def show_borrow_books(self):
        if not self.borrowed_books:
            print("No borrowed books.")
        else:
            print(f"{self.name}'s Borrowed Books:")
            for book in self.borrowed_books:
                print(f"- {book.title}")

    @staticmethod
    def is_valid_email(email):
        pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
        return re.match(pattern, email)

    @classmethod
    def len_member(cls):
        return cls.member_count


class Library:
    def __init__(self, name):
        self.name = name
        self.books = []
        self.members = []

    def add_book(self, book):
        self.books.append(book)
        print(f"'{book.title}' added successfully.")

    def remove_book(self, book_id):
        for book in self.books:
            if book.id == book_id:
                self.books.remove(book)
                Book.book_count -= 1
                print("Book removed.")
                return
        print("Book not found.")

    def show_all(self):
        print(f"\n===== {self.name} =====")
        print(f"Books   : {len(self.books)}")
        print(f"Members : {len(self.members)}")

    def search_books(self, keyword):
        found = False
        keyword = keyword.lower()

        for book in self.books:
            if keyword in book.title.lower() or keyword in book.author.lower():
                book.info()
                found = True

        if not found:
            print("No matching books found.")

    def add_member(self, member):
        self.members.append(member)
        print(f"{member.name} added successfully.")

    def borrow_book(self, member_id, book_id):
        member = None
        book = None

        for m in self.members:
            if m.id == member_id:
                member = m
                break

        for b in self.books:
            if b.id == book_id:
                book = b
                break

        if member and book:
            if book.borrow():
                member.borrow_book(book)
                print(f"{member.name} borrowed '{book.title}'.")
            else:
                print("Book is already borrowed.")
        else:
            print("Member or Book not found.")

    def return_book(self, member_id, book_id):
        member = None
        book = None

        for m in self.members:
            if m.id == member_id:
                member = m
                break

        for b in self.books:
            if b.id == book_id:
                book = b
                break

        if member and book:
            if book in member.borrowed_books:
                member.return_book(book)
                book.return_book()
                print(f"{member.name} returned '{book.title}'.")
            else:
                print("This member didn't borrow this book.")
        else:
            print("Member or Book not found.")

    def show_members(self):
        print("\nMembers:")
        for member in self.members:
            member.info()

    def show_books(self):
        print("\nBooks:")
        for book in self.books:
            book.info()

    def len_book(self):
        return len(self.books)

    def len_member(self):
        return len(self.members)


library = Library("City Library")

# Books
b1 = Book(1, "Python Basics", "John Smith", 20)
b2 = Book(2, "Data Structures", "Alice Brown", 30)
b3 = Book(3, "Machine Learning", "Andrew Ng", 50)

library.add_book(b1)
library.add_book(b2)
library.add_book(b3)

# Members
m1 = Member(1, "Alice", "alice@gmail.com")
m2 = Member(2, "Bob", "bob@yahoo.com")

library.add_member(m1)
library.add_member(m2)

library.show_all()

print("\nBorrowing Books")
library.borrow_book(1, 2)
library.borrow_book(2, 2)

m1.show_borrow_books()

print("\nReturning Book")
library.return_book(1, 2)

print("\nSearch Result")
library.search_books("Python")

print("\nAll Books")
library.show_books()

print("\nAll Members")
library.show_members()

print("Total Books:", Book.len_book())
print("Total Members:", Member.len_member())
```

```python
def square(num):
    return num **2

print(square(5))

    
```

```python
#List Comprehension
numbers=[1,2,3,4,5]
squared_number_lc=[num**2 for num in numbers]
squared_number_lc
```

```python
even_numbers=[num for num in range(1,11) if num%2==0]
even_numbers
```

```python
names=["John","Jane","Jim","Jack","Jill"]
upper_name=[name.upper() for name in names]
print(upper_name)
```

```python
name_contain=[name for name in names if name=="hari"]
print(name_contain)
```

```python
name_with_h=[name for name in names if name.startswith("H")]
print(name_with_h)
```

```python
numbers=[1,2,3,5,4,8,6]
squares=list(map(lambda x:x**2,numbers))
print(squares)
```

```python
numbers=[1,2,3,5,4,8,6]
even=list(filter(lambda x:x%2==0,numbers))
print(even)
```

Iterator vs Generator

```python
numbers=[10,20,30]
iterator=iter(numbers)
print(next(iterator))
print(next(iterator))
print(next(iterator))
```

```python
def count_upto(limit):
    number=1
    while number<=limit:
        yield number
        number+=1

for num in count_upto(10):
    print(num)
```

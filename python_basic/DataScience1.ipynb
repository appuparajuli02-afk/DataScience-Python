Student Management System

```python

students= [
   {
       'name':'Ram',
       'rollno': 1,
       'course': 'BICTE',
       'marks': 80
   },
   {
       'name': 'Sita',
       'rollno': 2,
       'course': 'BICTE',
       'marks': 85
   },
   {
       'name': 'Hari',
       'rollno': 3,
       'course': 'BICTE',
       'marks': 75
   },
   {
       'name': 'Laxman',
       'rollno':4,
       'course': 'BICTE',
       'marks': 70
   }
]

def welcome():
    print('*'*40)
    print("        WELCOME TO ABC CAMPUS         ")
    print('*'*40)
    # print(f"Welcome {students['name']}")

def show_menu():
    print('\n ---------ABC CAMPUS---------')
    print('1. Add Students')
    print('2. View Students')
    print('3. Search  Students')
    print('4. Update  Students')
    print('5. Delete  Students')
    print('6.Exit')

def add_students():
    name= input("Enter the students name: ")
    rollno= int(input("Enter the students rollno: "))
    course= input("Enter the students course: ")
    marks= float(input("Enter the students marks: "))

    student={
        "name":name,
        "rollno":rollno,
        "course":course,
        "marks":marks
      }

    students.append(student)
    print("\n Students added successfully")

def view_student():
    if len(students)==0:
        print("No students found")
        return

    print("-------Students List------")

    for student in students:
        print(f"Name:{student['name']}")
        print(f"Rollno:{student['rollno']}")
        print(f"Course:{student['course']}")
        print(f"Marks:{student['marks']}")
        print("*"*30)

def search_student():
    roll=int(input("Enter the rollno to search student:"))

    found=False

    for student in students:
        if student["rollno"]==roll:
            print("Student Found")
            print("-" *30)
            print(f"Student name:{student['name']}")
            print(f"Student rollno:{student['rollno']}")
            print(f"Student course:{student['course']}")
            print(f"Student marks:{student['marks']}")
            found=True
            break;

    if not found:
            print("Student not found")

def update_student():
    roll=int(input("Enter the rollno to update the student:"))

    for student in students:
        if student["rollno"]==roll:
            print("\n Leave blank if you dont want to change the value")

            name = input(f"Name ({student['name']}): ")
            if name:
                student["name"] = name

            course = input(f"Course({student['course']}): ")

            if course:
                student["course"] = course


            marks = (input(f"Marks({student['marks']}): "))
            if marks:
                student["marks"] = float(marks)


            print("Student updated successfully")
            return;

    print("Student not found")

def delete_student():
    roll=int(input("Enter the rollno of student to delete:"))

    for student in students:
        if student["rollno"]==roll:
            students.remove(student)
            print("Student deleted successfully")
            return;

    print("Student not found")

def menu():
    while True:
        show_menu()


        # print("\n===== STUDENT MANAGEMENT SYSTEM =====")
        # print("1. Add Student")
        # print("2. View Students")
        # print("3. Search Student")
        # print("4. Update Student")
        # print("5. Delete Student")
        # print("6. Exit")

        choice=input("Enter your choice:")

        if choice=='1':
            add_students()

        elif choice=='2':
            view_student()

        elif choice=='3':
            search_student()

        elif choice=='4':
            update_student()

        elif choice=="5":
            delete_student()

        elif choice=="6":
            print("Thank you for using the student management system")
            break 

        else:
            print("Invalid choice")

        
welcome()
menu()






```

```python
accounts=[
    {
        'Accno':123456789,
        'Name': 'Tony Stark',
        'Pin': 9876,
        'balance':15000
    },
    {
            'Accno':987654321,
            'Name': 'Steve Rogers',
            'Pin': 1234,
            'balance':10000
        },
    {
                'Accno':984524310,
                'Name': 'Sam Wilson',
                'Pin': 5678,
                'balance':20000
     },
     {
             'Accno':984506677,
             'Name': 'Peter Parker',
             'Pin': 9012,
             'balance':5000
         },
    {
            'Accno':751111223,
            'Name': 'Wanda',
            'Pin': 8520,
            'balance':25000
        }
]

def welcome():
    print('*'*30)
    print("Welcome to XYZ Bank")
    print('*'*30)

def verify_pin():
    acc=int(input("Enter Account Number:"))
    entered_pin=int(input("Enter your pin"))
    for account in accounts:
        if account['Accno']==acc and account['Pin']== entered_pin:
          print("Login Successfully")
          return account
    
    print("Incorrect account number or Pin")
    return None

def show_menu():
    print("\n-------Bank Management System Menu---------")
    print("1.Create new account")
    print("2.View All Account")
    print("3. Search Account")
    print("4.Deposit Money")
    print("5.Withdraw Money")
    print("6.Change PIN")
    print("7.Delete Account")
    print("8.Exit")

def create_account():
    accno=input("Enter your account numbber:")
    name=input("Enter your name:")
    pin=int(input("Enter your name:"))
    balance=float(input("Enter your balance:"))

    account={
        "Accno": accno,
        "Name":name,
        "Pin":pin,
        "balance":balance
         }
    accounts.append(account)
    print("\n New Account Created Successfully")

def view_account():
    if len(accounts)==0:
        print("No account found")
        return
    print("------Accounts Details-------")

    for account in accounts:
        print(f"AccNo: {account['Accno']}")
        print(f"Name: {account['Name']}")
        print(f"Pin: {account['Pin']}")
        print(f"Balance: {account['balance']}")
        print('*'*30)

def search_accounts():
    acc=int(input("Enter your account number:"))
    found=False
    for account in accounts:
        if account['Accno']==acc:
            print("Account found")
            print("*"*30)
            print(f"Accno:{account['Accno']}")
            print(f"Name:{account['Name']}")
            print(f"Balance:{account['balance']}")
            found=True
            break;

    if not found:
            print("Account not found")

def deposit(account):
    amount=float(input("Enter the money you want to deposit:"))
    if amount>0:
        account['balance']+=amount
        print("Deposit Successfull")
        print("New Balance:",account['balance'])
    else:
        print("Invalid Amount")

def withdraw(account):
    amount=float(input("Enter the money you want to withdraw:"))
    if amount<=0:
        print("Invalid Amount")
    elif amount> account['balance']:
        print("Insufficient balance")
    else:
        account['balance']=account['balance']-amount
        print("Withdraw successfull")
        print("New Balance:",account['balance'])

def change_pin(account):
    old_pin=int(input("Enter your old pin:"))
    if old_pin==account['Pin']:
        new_pin=int(input("Enter your new pin:"))
        account['Pin']=new_pin
        print("Pin changed successfully")
    else:
        print("Incorrect Pin")

def delete_account(account):
    # acc=int(input("Enter your account number:"))
    # for account in accounts:
    #     if account['Accno']==acc:
    #         accounts.remove(account)
    #         print("Account deleted successfully")
    #         return;
    # print("Account not found")
    accounts.remove(account)
    print("Account deleted successfully")

def menu():
    current_account =verify_pin()
    if current_account is None:
        return
    while True:
        show_menu()
        choice=input("Enter your choice:")

        if choice=='1':
            create_account()

        elif choice=='2':
            view_account()

        elif choice=='3':
            search_accounts()

        elif choice=='4':
            deposit(current_account)

        elif choice=='5':
            withdraw(current_account)

        elif choice=='6':
            change_pin(current_account)

        elif choice=='7':
            delete_account(current_account)

        elif choice=='8':
            print("Thank you for choosing our XYZ Bank")
            break
        else:
            print("Invalid Choice")

welcome()
menu()





        






```

Expense Tracker

```python
expenses=[
    {
        'Expense_id':1,
        'Title': 'Lunch',
        'Amount':300,
        'Category':'Food'
    },
    {
        'Expense_id':2,
        'Title':'Movies',
        'Amount':700,
        'Category':'Entertainment'
    },
    {
        'Expense_id':3,
        'Title':'Snacks',
        'Amount':1500,
        'Category':'Food'
    },
    {
        'Expense_id':4,
        'Title':'Clothes',
        'Amount':3000,
        'Category':'Clothing'
    }
]

def welcome():
    print('='*30)
    print("Welcome to Expense Tracker")
    print('='*30)

def show_menu():
    print("\n -------EXPENSE TRACKER--------")
    print("1. Add Expense")
    print("2. View Expense")
    print("3. Search Expense")
    print("4. Delete Expense")
    print("5. Total Expense")
    print("6. Exit")

def add_expense():
    id=int(input("Enter your expense id:"))
    title=input("Enter your expense title:")
    amount=float(input("Enter your expense amount:"))
    category=input("Enter your expense category:")

    expense={
        "Expense_id":id,
        "Title":title,
        "Amount":amount,
        "Category":category
    }
    expenses.append(expense)
    print("New expense added successfully")

def view_expense():
    if len(expenses)==0:
        print("No expenses found")
        return
    print("Expenses details")

    for expense in expenses:
        #print(f"Expense ID:{expense['expense_id']}")
        print(f"Title:{expense['Title']}")
        print(f"Amount:{expense['Amount']}")
        print(f"Category:{expense['Category']}")
        print('='*30)

def search_expense():
    id=int(input("Enter your expense id:"))
    found=False
    for expense in expenses:
        if expense['Expense_id']==id:
            print("Expense found:")
            print('='*30)
            print(f"Title:{expense['Title']}")
            print(f"Amount:{expense['Amount']}")
            print(f"Category:{expense['Category']}")
            found=True
            break
    if not found:
        print("Expense not found")

def delete_expense():
    id=int(input("Enter your expense id:"))
    for expense in expenses:
        if expense['Expense_id']==id:
            expenses.remove(expense)
            print("Expense deleted")
            return 
    print("Expense not found found")

def total_expense():
    total=0
    for expense in expenses:
        total+=expense['Amount']

    print("Total Expense is:",total)

def menu():
     while True:
            show_menu()
            choice=input("Enter your choice:")
    
            if choice=='1':
                add_expense()
    
            elif choice=='2':
                view_expense()
    
            elif choice=='3':
                search_expense()
    
            elif choice=='4':
                delete_expense()
    
            elif choice=='5':
                total_expense()
    
        
    
            elif choice=='6':
                print("Thank you for choosing to track expense")
                break
            else:
                print("Invalid Choice")

welcome()
menu()
    
    
    
```

Python Modules


```python
while True:
    # 1. Show the menu
    print("\n=== Diary Menu ===")
    print("1. Write Diary")
    print("2. Read Diary")
    print("3. Exit")
    
    # 2. Get the user's choice
    choice = input("Choose an option (1, 2, or 3): ")
    
    # 3. Handle the choice
    if choice == "1":
        entry = input("Type your diary entry: ")
        
        # Open file in 'a' (append) mode so it adds to the bottom
        with open  ("diary.txt", "w") as file :
            file.write("- " + entry + "\n")
           
        
        print("Saved to diary.txt!")
        
    elif choice == "2":
        try:
            # Open file in 'r' (read) mode
            with  open("diary.txt", "r") as file:
                print("\n--- Your Diary ---")
                print(file.read())
                
            
        except FileNotFoundError:
            # If the file hasn't been created yet
            print("Your diary is empty! Try writing something first.")
            
    elif choice == "3":
        print("Goodbye!")
        file.close()
        break # This stops the while loop
        
    else:
        print("Wrong choice, please try again.")
```

Practicing exception handling

```python
try:
    num1=int(input("Enter the first number:"))
    num2=int(input("Enter the second number:"))
    result=num1/num2
except ZeroDivisionError:
    print("Cannot be divided by zero!")
except ValueError:
    print("Please enter a valid number.")

else:
    print("The result is:",result)

finally:
    print("Program ended.")
```

```python
try:
    age=int(input("Enter your age:"))
except ValueError:
    print("Please enter integer number only.")
else:
    print("Your age is:",age)

finally:
    print("Program ended.")
```

apply()

```python
employee_df["salary"] = pd.to_numeric(employee_df["salary"])
```

```python
def salary_level(salary):
    if salary>65000:
        return "High"
    elif salary >50000:
        return"Medium" 
    else:
        return "Low"

```

```python
employee_df["salary_level"]=employee_df["salary"].apply(salary_level)
employee_df
```

```python
#lamda version
# employee_df=employee_df["salary"]=employee_df["salary"].apply(lambda x: "High" if x>=60000 else "Low")
# employee_df.head()

employee_df["salary_level"] = employee_df["salary"].apply(
    lambda x: "High" if x >= 60000 else "Low"
)

employee_df.head()
```

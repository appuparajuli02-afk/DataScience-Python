19. Create a salary_level column:
```python
< 50,000 → Low
50,000–80,000 → Medium
> 80,000 → High

````python
def salary_level(salary):
    if salary<50000:
        return "Low"
    elif salary<=80000:
        return "Medium"
    else:
        return"High" 

employee_df["salary_level"]=employee_df["salary"].apply(salary_level)
````

````python
print(employee_df[["salary", "salary_level"]])
````

20. Create an experience_level column such as Junior, Mid, and Senior based on years of experience.

````python
def experience_level(experience_years):
    if experience_years<=2:
        return "Junior"
    elif experience_years<=5:
        return "Mid"
    else:
        return "Senior"

employee_df["experience_level"]=employee_df["experience_years"].apply(experience_level)
````

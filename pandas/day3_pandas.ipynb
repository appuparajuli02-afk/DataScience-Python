````python

import numpy as np
import pandas as pd

data = {
    "name": [
        "   Ram", "Sita", "Hari", "Gita",
        "NaBin", "ANita", "Bikash", "    Sunita"
    ],
    "department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "HR"
    ],
    "age": [25, 28, 32, 30, 26, 35, 29, 31],
    "salary": [
        50000, 65000, 75000, 70000,
        55000, 80000, 68000, 62000
    ]
}

employees = pd.DataFrame(data)
employees.head()
````

````python
def salary_category(salary):
    if salary < 60000:
        return "Low"
    elif salary <= 70000:
        return "Medium"
    else:
        return "High"
````

````python
employees["salary_category"] = employees["salary"].apply(salary_category)
employees
````

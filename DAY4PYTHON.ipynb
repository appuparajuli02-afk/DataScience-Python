```python
def max_number(numbers):
  largest=numbers[0]

  for num in numbers:
    if num>largest:
      largest=num
  return largest
numbers=[10,20,30,4,8,9,39]
print("The largest number is", max_number(numbers))
```

```python
accounts=[
    {
        'Accno':123456789,
        'Name': 'Tony Stark',
        'Pin': 9876,
        'balance':15000
    },
    {
            'Accno':987654321,
            'Name': 'Steve Rogers',
            'Pin': 1234,
            'balance':10000
        },
    {
                'Accno':984524310,
                'Name': 'Sam Wilson',
                'Pin': 5678,
                'balance':20000
     },
     {
             'Accno':984506677,
             'Name': 'Peter Parker',
             'Pin': 9012,
             'balance':5000
         },
    {
            'Accno':751111223,
            'Name': 'Wanda',
            'Pin': 8520,
            'balance':25000
        }
]

def welcome():
    print('*'*30)
    print("Welcome to XYZ Bank")
    print('*'*30)

def verify_pin():
    acc=int(input("Enter Account Number:"))
    entered_pin=int(input("Enter your pin"))
    for account in accounts:
        if account['Accno']==acc and account['Pin']== entered_pin:
          print("Login Successfully")
          return account

    print("Incorrect account number or Pin")
    return None

def show_menu():
    print("\n-------Bank Management System Menu---------")
    print("1.Create new account")
    print("2.View All Account")
    print("3. Search Account")
    print("4.Deposit Money")
    print("5.Withdraw Money")
    print("6.Change PIN")
    print("7.Delete Account")
    print("8.Exit")

def create_account():
    accno=input("Enter your account numbber:")
    name=input("Enter your name:")
    pin=int(input("Enter your name:"))
    balance=float(input("Enter your balance:"))

    account={
        "Accno": accno,
        "Name":name,
        "Pin":pin,
        "balance":balance
         }
    accounts.append(account)
    print("\n New Account Created Successfully")

def view_account():
    if len(accounts)==0:
        print("No account found")
        return
    print("------Accounts Details-------")

    for account in accounts:
        print(f"AccNo: {account['Accno']}")
        print(f"Name: {account['Name']}")
        print(f"Pin: {account['Pin']}")
        print(f"Balance: {account['balance']}")
        print('*'*30)

def search_accounts():
    acc=int(input("Enter your account number:"))
    found=False
    for account in accounts:
        if account['Accno']==acc:
            print("Account found")
            print("*"*30)
            print(f"Accno:{account['Accno']}")
            print(f"Name:{account['Name']}")
            print(f"Balance:{account['balance']}")
            found=True
            break;

    if not found:
            print("Account not found")

def deposit(account):
  while True:
    try:
        amount=float(input("Enter the money you want to deposit:"))
        if amount<=0:
              print("Invalid Amount! Please enter a positive number.")
        else:
             account['balance']+=amount
             print("Deposit Successfull")
             print("New Balance:",account['balance'])
             break
    except ValueError:
          print("Please enter a valid number.")

def withdraw(account):
  try:
    amount=float(input("Enter the money you want to withdraw:"))
    if amount<=0:
        print("Invalid Amount")

    elif amount> account['balance']:
        print("Insufficient balance")
    else:
        account['balance']=account['balance']-amount
        print("Withdraw successfull")
        print("New Balance:",account['balance'])
  except ValueError:
        print("Invalid Amount")

def change_pin(account):
    old_pin=int(input("Enter your old pin:"))
    if old_pin==account['Pin']:
        new_pin=int(input("Enter your new pin:"))
        account['Pin']=new_pin
        print("Pin changed successfully")
    else:
        print("Incorrect Pin")

def delete_account(account):
    # acc=int(input("Enter your account number:"))
    # for account in accounts:
    #     if account['Accno']==acc:
    #         accounts.remove(account)
    #         print("Account deleted successfully")
    #         return;
    # print("Account not found")
    accounts.remove(account)
    print("Account deleted successfully")

def menu():
    current_account =verify_pin()
    if current_account is None:
        return
    while True:
        show_menu()
        choice=input("Enter your choice:")

        if choice=='1':
            create_account()

        elif choice=='2':
            view_account()

        elif choice=='3':
            search_accounts()

        elif choice=='4':
            deposit(current_account)

        elif choice=='5':
            withdraw(current_account)

        elif choice=='6':
            change_pin(current_account)

        elif choice=='7':
            delete_account(current_account)

        elif choice=='8':
            print("Thank you for choosing our XYZ Bank")
            break
        else:
            print("Invalid Choice")

welcome()
menu()
```

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero."
    return a / b


def calculator():
    while True:
        print("\n===== SIMPLE CALCULATOR =====")
        print("1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ")

        if choice == "5":
            print("Thank you for using the calculator!")
            break

        if choice not in ["1", "2", "3", "4"]:
            print("Invalid choice! Please try again.")
            continue

        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))

            if choice == "1":
                print("Result:", add(num1, num2))

            elif choice == "2":
                print("Result:", subtract(num1, num2))

            elif choice == "3":
                print("Result:", multiply(num1, num2))

            elif choice == "4":
                print("Result:", divide(num1, num2))

        except ValueError:
            print("Please enter valid numbers.")


calculator()
```

Student Grade Calculator

```python
def calculate_grade(marks):
  if marks>=90:
    return "A+"
  elif marks>=80:
    return "A"
  elif marks>=70:
    return "B+"
  elif marks>=60:
    return "B"
  elif marks>=50:
    return "C"
  elif marks>=40:
    return "D"
  else:
    return "E"

def student():
  try:
   name=input("Enter the name of student:")
   marks=float(input("Enter the marks of the student:"))

   if marks<0 or marks>100:
     print("Marks should be between 0 and 100")
     return

   grade=calculate_grade(marks)
   print("----Result-----")
   print("Name:",name)
   print("Marks:",marks)
   print("Grade:",grade)

  except ValueError:
   print("Please enter a valid number.")

student()

```

Voting Eligibility

```python
def check_eligibility(age):
  if age>=18:
    return "You are eligible to vote."
  elif age>=0:
    return "You are not eligible to vote."
  else :
    return "Age cannot be in negative."

def voting():
  try:
    name=input("Enter you name:")
    age=int(input("Enter your age:"))

    result=check_eligibility(age)
    print("-----Result-----")
    print("Name:",name)
    print("Age:",age)
    print("Result:",result)

  except ValueError:
    print("Enter a valid age.")

voting()
```

```python
numbers = [1, 2, 3, 4, 5, 8]
new_numbers = []

for i in numbers:
    new_num = i ** 2
    new_numbers.append(new_num)

print(new_numbers)
```

```python
numbers=[50,90,80,20,45]
largest_number=numbers[0]

for num in numbers:
    if num> largest_number:
        largest_number=num

print("The largest number is:",largest_number)
```

```python
word="banana"
frequency=[]

for char in word:
    if char in frequency:
     frequency[char]+=1

    else:
       frequency[char]=1

print(frequency)


```

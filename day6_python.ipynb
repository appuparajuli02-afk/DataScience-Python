```python
# banking system
class Bank:
  bank_name = "NIC Asia"

  # constructor (called automatically when object is created)
  def __init__(self, name, address, balance, pin):
    self.name = name
    self.address = address
    self.balance = balance

  def deposit(self, amount):
    if not str(amount).isdigit() or amount <= 0:
      return "invalid amount"

    self.balance += amount
    return self.balance

  def withdraw(self, amount):
    if not str(amount).isdigit() or amount <= 0:
      return "invalid amount"

    if amount > self.balance:
      return "insufficient balance"

    self.balance -= amount
    return self.balance

  def check_balance(self):
    return self.balance
```

```python
# Example

# parent class
class Animal:

  def eat(self):
    print("Animal is eating...")

# child class
class Dog(Animal):
  def bark(self):
    print("Dog is barking...")

class Cat(Animal):
  def meow(self):
    print("Cat is meowing...")

class Tiger(Animal):
  def roar(self):
    print("Tiger is roaring...")
```

```python
class Person:
  def __init__(self,name,age,gender,address):
    self.name=name
    self.age=age
    self.gender=gender
    self.address=address

  def display_info(self):
    return f"The person name is {self.name}, age is {self.age}, gender is {self.gender} and lives in {self.address}"

class Employee(Person):
  company="ABC limited"
  def __init__(self,name,age,gender,address,salary,post,department):
    super().__init__(name,age,gender,address)
    self.salary=salary
    self.post=post
    self.department=department

  def display(self):
    super().display_info()
    return f"Salary is {self.salary}, post is {self.post} and department is {self.department}"

  def increase_salary(self,amount):
    self.salary += amount
    return self.salary

  def change_dept(self,new_dept):
    self.department=new_dept
    return new_dept

  def annual_salary(self):
    return self.salary*12



e1 = Employee("Hari", 25, "Male", "Pokhara", 50000, "Manager", "IT")

# print(e1.display_info())
print(e1.display())
print("Annual Salary:", e1.annual_salary())

e1.increase_salary(5000)
print("New Salary:", e1.salary)

e1.change_dept("HR")
print("New Department:", e1.department)

```

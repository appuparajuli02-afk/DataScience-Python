```code

    [65, 70, 75],
    [80, 85, 90],
    [45, 55, 60],
    [90, 95, 92],
    [35, 40, 50],

    [88, 82, 84],
    [55, 60, 58]

```
columns: Python | Mathematics | Statistics
 
### calculate
1. Average mark of every student
2. Average mark of every subject
3. Highest mark in the dataset
4. Lowest mark
5. Student with highest average
6. Students whose Python mark is above 70
7. Number of individual marks above 80
8. Students with an overall average below 50
9. Sort students based on overall average
10. Assign a result using:
```code
Average >= 80     → Distinction
Average >= 60     → First Division
Average >= 40     → Pass
Average < 40      → Fail
 

````python
marks = np.array([
    [65, 70, 75],
    [80, 85, 90],
    [45, 55, 60],
    [90, 95, 92],
    [35, 40, 50],
    [72, 68, 75],
    [88, 82, 84],
    [55, 60, 58]
])

#Average marks of every student
average_marks=np.average(marks,axis=1)
print("The average marks of every students is:\n",average_marks)
#print("The average marks of each students is:\n",np.average(marks,axis=1))

#Average marks of every subject
print("The average marks of every subject is:\n",np.average(marks,axis=0))

#Highest marks
print("The highest marks in database is:",np.max(marks))

#Lowest mark
print("The lowest mark in database is:",np.min(marks))

#Student with highest average
print("The student with highest average is:",np.max(average_marks))

# python Marks above 70
print("The students marks of python above 70 are:\n",marks[marks[:,0]>70,0])

#Marks above 80
print("The individual whose marks are above 80 are\n:",marks[marks>80])

#Average below 50:
print("The individual whose marks are below 50 are\n:",average_marks[average_marks<50])

#sort based on average:
print("The sorted order based on the average marks is:\n",np.sort(average_marks))

#Assign a result:
#new_salary=np.where(salary<=4000, salary*1.2, salary*1.1)

result=np.select(
                [average_marks>=80,
                average_marks>=60,
                average_marks>=40],

                [
                    "Distinction",
                    "First division",
                    "Second division"
                ] ,
                default='fail'

                )
print(result)
````

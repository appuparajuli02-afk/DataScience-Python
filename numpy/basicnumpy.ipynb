```python
import numpy as np

temp = np.array([21, 25, 28, 30, 27, 24, 22])

# Average
avg_temp = np.average(temp)
print("The average temperature is:", avg_temp)

# Maximum
print("The maximum temperature is:", np.max(temp))

# Minimum
print("The minimum temperature is:", np.min(temp))

# Temperature above average
print("The temperature above average is:", temp[temp > avg_temp])
```

Sales Analysis
- Print the array dimensions.
- Print its shape.
- Find total sales.
- Find average sales.
- Find the highest sale.
- Find the lowest sale.
- Calculate total sales for each salesperson.
- Calculate total sales for each month.
- Find all sales greater than 1500.
- Increase every sales value by 10% and print the resulting array.

```python
sales = np.array([
    [1200, 1500, 1100],
    [1800, 1700, 1600],
    [900, 1200, 1400],
    [2000, 2100, 1900]
])

print(sales.ndim)
print(sales.shape)
#Total sales
print("The total sales is:",np.sum(sales))

#Average sales
print("The average sales is:",np.average(sales))

#Highest sale
print("The highest sale is:",np.max(sales))

#Lowest sale
print("The lowest sale is:",np.min(sales))

#Sales by each person
print("The sales of first salesperson is:",np.sum(sales[0]))

print("The sales of second salesperson is:",np.sum(sales[1]))

print("The sales of third salesperson is:",np.sum(sales[2]))

print("The sales of fourth salesperson is:",np.sum(sales[3]))

#sales by each month  for columns
print("The total sales of Baisakh is:",np.sum(sales,axis=0))

#sales by each salesperson: for rows 
print("The total sales of each salesperson is:",np.sum(sales,axis=1))

#sales greater than 1500
result=sales[sales>1500]
print("The sales greater than 1500 are:",result)

#Increase by 10 percent
new_sales=sales*1.1
print("Every sales increase by 10 percentage is:\n",new_sales)





```
